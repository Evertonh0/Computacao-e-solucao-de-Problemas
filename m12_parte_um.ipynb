### **Parte 1 | Desafio**
Com base no que você estudou até o momento, desenvolva um código-fonte em Python (*.py) que solicite para o usuário digitar uma sequência numérica de 11 dígitos inteiros que, supostamente, representem um número de documento CPF. Considere que os dados de entrada serão compostos apenas por números (o usuário não deverá digitar pontos e/ou traços).

Este código, desenvolvido por você, deverá ser capaz de verificar se a sequência numérica digitada pelo usuário é um CPF válido. A checagem deverá considerar, para a tomada de decisão, se os dígitos verificadores digitados (os últimos dois dígitos) são coerentes.

Caso o CPF seja válido, o código deverá imprimir uma mensagem indicando que a sequência numérica é válida, conforme apresentado na imagem abaixo. Na hipótese do CPF digitado ser inválido, isto é, dos dígitos verificadores digitados serem diferentes dos calculados pelo algoritmo, o código deverá imprimir uma mensagem indicando que o CPF é inválido.

```python
cpf_input = input("Digite a sequencia numerica de 11 digitos do CPF")

if len(cpf_input) != 11:
    print("O CPF digitado e invalido.")
else:
    digitos = []

    for letra in cpf_input:
        digitos.append(int(letra))

    todos_iguais = True
    for d in digitos:
        if d != digitos[0]:
            todos_iguais = False
            break

    if todos_iguais:
        print("O CPF digitado e invalido.")
    else:
        soma1 = 0
        multiplicador1 = 10

        for i in range(9):
            soma1 = soma1 + (digitos[i] * multiplicador1)
            multiplicador1 = multiplicador1 - 1

        resto1 = soma1 % 11

        if resto1 < 2:
            digito1_calculado = 0
        else:
            digito1_calculado = 11 - resto1

        soma2 = 0
        multiplicador2 = 11

        for i in range(10):
            soma2 = soma2 + (digitos[i] * multiplicador2)
            multiplicador2 = multiplicador2 - 1

        resto2 = soma2 % 11

        if resto2 < 2:
            digito2_calculado = 0
        else:
            digito2_calculado = 11 - resto2

        if digitos[9] == digito1_calculado and digitos[10] == digito2_calculado:
            print("A sequencia numerica e valida.")
        else:
            print("O CPF digitado e invalido.")
```
