Problema 1

```python

num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))
maior_numero = num1

if num2 > maior_numero:
    maior_numero = num2


if num3 > maior_numero:
    maior_numero = num3

print("O maior número é:", maior_numero)
```

Problema 3

```python

peso = float(input("Digite seu peso em kg (ex: 70.5): "))
altura = float(input("Digite sua altura em metros (ex: 1.75): "))


imc = peso / (altura ** 2)


print(f"Seu IMC é: {imc:.2f}")

# Classifica o IMC
if imc < 18.5:
    print("Classificação: Baixo peso")
elif imc < 25.0:
    print("Classificação: Normal")
elif imc < 30.0:
    print("Classificação: Excesso de peso")
elif imc < 35.0:
    print("Classificação: Obesidade classe 1")
elif imc < 40.0:
    print("Classificação: Obesidade classe 2")
else:
    print("Classificação: Obesidade classe 3")
```
