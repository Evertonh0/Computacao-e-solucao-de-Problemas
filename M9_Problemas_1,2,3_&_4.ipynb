1) Crie um programa que imprima a contagem regressiva de 10 até 1 e, em seguida, imprima a mensagem "Feliz Ano Novo!".

```python
i = 10
while i >= 1:
    print(i)
    i = i - 1
print("Feliz Ano Novo!")
```

2) Faça um programa que receba dois números inteiros e imprima os números inteiros que estão no intervalo compreendido por eles.

```python
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
if num1 < num2:
    inicio = num1 + 1
    fim = num2
else:
    inicio = num2 + 1
    fim = num1
while inicio < fim:
    print(inicio)
    inicio = inicio + 1
```

3) Faça um programa que solicite ao usuário um número (maior que 1) e imprima todos os números pares de 1 até esse número escolhido pelo usuário.

```python
numero = int(input("Digite um número maior que 1: "))
contador = 2
while contador <= numero:
    print(contador)
    contador = contador + 2
```

4) Faça um programa que calcule o fatorial de um número inteiro fornecido pelo usuário. Exemplo: 5! = 5.4.3.2.1 = 120.


```python
num = int(input("Digite um número inteiro: "))
fatorial = 1
i = 1
if num < 0:
    print("Fatorial não pode ser calculado para números negativos.")
elif num == 0:
    print("O fatorial de 0 é 1")
else:
    while i <= num:
        fatorial = fatorial * i
        i = i + 1
    print(fatorial)
```
