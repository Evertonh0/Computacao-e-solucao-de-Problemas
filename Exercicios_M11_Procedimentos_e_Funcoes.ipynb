### **Desafio 1)** Escreva uma função chamada “par_ou_impar” que receba um número inteiro e retorne a “string” "par" se o número for par, e "ímpar" caso contrário.

```python
def par_ou_impar(numero):
  if numero % 2 ==0:
    return "par"
  else:
    return "impar"

dado_usuario = int(input("Digite um número inteiro"))

resultado_def_par_ou_impar = par_ou_impar(dado_usuario)

print(f"O número {dado_usuario} é {resultado_def_par_ou_impar}")

```

### **Desafio 2)** Implemente uma função chamada “fatorial” que receba um número inteiro não negativo e retorne o fatorial desse número.

```python
def fatorial(numero):
    if numero == 0:
        return 1

    resultado = 1
    for i in range(1, numero + 1):
        resultado *= i

    return resultado

try:
    dado_usuario = int(input("Digite um número inteiro e não negativo: "))

    if dado_usuario < 0:
        print("Número digitado é menor que zero!")
    else:
        resp = fatorial(dado_usuario)
        print(f"O fatorial de {dado_usuario} é {resp}")

except ValueError:
    print("Erro! Digite um número válido!!")
```

### **Desafio 3)** Escreva uma função chamada “eh_primo” que receba um número inteiro e retorne “true” se ele for primo, e “false” caso contrário.

```python
def eh_primo(numero):
  if numero <= 1:
    return False
  for i in range(2,numero):
    if numero % i == 0:
      return False
  return True

try:
  dado_usuario = int(input("Digite um numero inteiro pra ver se ele é primo: "))
  resultado = eh_primo(dado_usuario)
  if resultado:
        print(f"O número {dado_usuario} é primo! (True)")
  else:
        print(f"O número {dado_usuario} NÃO é primo. (False)")

except ValueError:
    print("Erro! digite um número inteiro válido.")
```

### **Desafio 4)** Crie uma função chamada “fibonacci” que receba um número inteiro “n” e retorne uma lista com os “n” primeiros termos da sequência de Fibonacci.

```python
def fibonacci(n):
    # 1. Tratamento para quantidades menores ou iguais a zero
    if n <= 0:
        return []

    # 2. Se n for igual a 1, retornamos apenas o primeiro termo
    if n == 1:
        return [0]

    # 3. Lista inicial com os dois primeiros termos da sequência
    fib = [0, 1]

    # 4. Usamos um loop para gerar os termos restantes
    # Como já temos 2 termos, precisamos calcular (n - 2) termos adicionais
    for _ in range(n - 2):
        # Soma os dois últimos elementos da lista atual
        proximo_termo = fib[-1] + fib[-2]
        # Adiciona o novo termo no final da lista
        fib.append(proximo_termo)

    return fib

# --- Interagindo com o usuário no terminal ---
try:
    quantidade = int(input("Digite quantos termos de Fibonacci você deseja gerar: "))

    if quantidade < 0:
        print("Por favor, digite um número inteiro não negativo.")
    else:
        # Chamamos a função e guardamos a lista retornada
        resultado_fib = fibonacci(quantidade)
        print(f"Os {quantidade} primeiros termos de Fibonacci são: {resultado_fib}")

except ValueError:
    print("Erro! Digite um número inteiro válido.")
```

### **Desafio 5)** Escreva uma função chamada "contar_vogais” que receba uma “string” e retorne à quantidade de vogais presentes na “string”.

```python
def contar_vogais(texto):

    vogais = "aeiouAEIOU"
    contador = 0


    for letra in texto:

        if letra in vogais:
            contador += 1

    return contador


frase_usuario = input("Digite uma palavra ou frase: ")
quantidade_encontrada = contar_vogais(frase_usuario)


print(f"A string digitada possui {quantidade_encontrada} vogais.")
```

### **Desafio 6)** Crie uma função chamada “celsius_para_fahrenheit” que receba uma temperatura em graus Celsius e retorne a temperatura equivalente na escala Fahrenheit. A expressão de conversão entre as escalas é:

```python
def celsius_para_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

try:
    temp_celsius = float(input("Digite a temperatura em graus Celsius: "))
    temp_fah = celsius_para_fahrenheit(temp_celsius)

    print(f"{temp_celsius}°C é igual a {temp_fah:.2f}°F")

except ValueError:
    print("Erro! digite um valor válido.")
```

### **Desafio 7)** Crie uma função chamada “calcular_area” que receba dois parâmetros:

* largura (obrigatório), um número que representa a largura do retângulo.
* altura (opcional), um número que representa a altura, cujo valor padrão é o mesmo da largura (caso seja um quadrado).
A função deve retornar a área do retângulo ou quadrado.

```python
def calcular_area(largura, altura=None):

    if altura is None:
        altura = largura
    area = largura * altura
    return area

try:
    print("Calculadora de Área")
    l = float(input("Digite a largura: "))

    h_input = input("Digite a altura: ").strip()

    if h_input == " ":
        resultado = calcular_area(l)
        print(f"area do quadrado com lado {l} é {resultado}")
    else:
        h = float(h_input)
        resultado = calcular_area(l, h)
        print(f" area do retângulo de {l} x {h} é: {resultado}")

except ValueError:
    print("Erro! digite valores numéricos válidos.")
```

### **Desafio 8)** Crie uma função chamada “calcular_juros_simples” que receba três parâmetros:

* capital (obrigatório), o valor inicial do capital.
* taxa (obrigatório), a taxa de juros mensal em porcentagem.
* tempo (opcional), o tempo em meses, cujo valor padrão é 1 mês.

A função deve retornar o valor dos juros simples, calculado pela expressão:

```python
def calcular_juros_simples(capital, taxa, tempo=1):
    return (capital * taxa * tempo) / 100

try:
    print("Calculadora de Juros Simples")

    cap = float(input("Digite o valor do capital inicial: "))
    tx = float(input("Digite a taxa de juros mensal (ex: 5 para 5%): "))
    tempo_input = input("Digite o tempo em meses: ").strip()

    if tempo_input == "":

        resultado = calcular_juros_simples(cap, tx)
        print(f"\nO valor dos juros simples para 1 mês é: R$ {resultado:.2f}")
    else:

        tp = float(tempo_input)
        resultado = calcular_juros_simples(cap, tx, tp)
        print(f"\nO valor dos juros simples para {tp} meses é: R$ {resultado:.2f}")

except ValueError:
    print("Erro! digite apenas valores numéricos válidos.")
```
