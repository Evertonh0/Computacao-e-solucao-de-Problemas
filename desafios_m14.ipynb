### **Desafio 1: Considere um dicionário para organizar os dados de um acervo musical, como uma coleção de LPs. Algumas possíveis chaves para esse acervo incluem: artista, título do LP, quantidade de faixas e duração total (em minutos)**.

Partindo dessa premissa:

Desenvolva uma função para criar um registro na coleção, que deve ler os valores das chaves a partir do teclado e retornar uma variável do tipo dict. Essa função não precisa ter argumentos de entrada e deve poder ser usada sempre que for necessário adicionar um registro a uma lista, chamada de acervo.
Crie um programa que insere registros utilizando sua função e que deve, ao final, mostrar todo o acervo.

```python
def adicionar_registro():

  artista = input('Digite o nome do artista: ')
  titulo = input('Digite o titulo da obra: ')
  quantidade = int(input('Digite quantas músicas tem: '))
  duracao_total = float(input('Digite a duração total (em minutos): '))

  registro = {

    'artista': artista,
    'titulo': titulo,
    'quantidade_faixas': quantidade,
    'duracao_total': duracao_total
  }
  return registro


quantidade_lps = int(input('Quantos LPs deseja registrar? '))

acervo = [adicionar_registro() for informacao_LP in range(quantidade_lps)]

print('Acervo musical adicionado:')
for lp in acervo:
  print(lp)
```

### **Desafio 2: Crie um programa que leia uma lista com 5 palavras a partir do teclado e, a partir dessa lista, crie duas outras listas: uma com as palavras que contêm 6 ou mais letras e outra com as palavras que contêm menos do que 6 letras.**

```python
palavras = [input(f'Digite a palavra {i + 1}: ') for i in range(5)]

palavras_grandes = [p for p in palavras if len(p) >= 6]
palavras_pequenas = [p for p in palavras if len(p) < 6]

print(f'Lista original: {palavras}')
print(f'Palavras com 6 ou mais letras: {palavras_grandes}')
print(f'Palavras com menos de 6 letras: {palavras_pequenas}')
```
